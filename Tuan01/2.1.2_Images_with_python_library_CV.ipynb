First, let's define a helper function to concatenate two images side-by-side. You will need to understand this code this moment, but this function will be used repeatedly in this tutorial to showcase the results.


```python
from tkinter import Image  # import Image từ tkinter
def get_concat_h(im1, im2):  # hàm nối 2 ảnh theo chiều ngang
    #https://note.nkmk.me/en/python-pillow-concat-images/
    dst = Image.new('RGB', (im1.width + im2.width, im1.height))  # tạo ảnh mới với kích thước = tổng chiều rộng 2 ảnh
    dst.paste(im1, (0, 0))  # dán ảnh đầu tiên vào vị trí (0,0)
    dst.paste(im2, (im1.width, 0))  # dán ảnh thứ 2 vào bên phải ảnh đầu
    return dst  # trả về ảnh đã nối
```

## Image Files and Paths  


An image is stored as a file on your computer. Below, we define `my_image` as the filename of a file in this directory.


```python
my_image = "lenna.png"  # tên file ảnh
```

Filename consists of two parts, the name of the file and the extension, separated by a full stop (`.`). The extension specifies the format of the image. There are two popular image formats -- Joint Photographic Expert Group image (or `.jpg`, `.jpeg`) and Portable Network Graphics (or `.png`). These file types make it simpler to work with images. For example, it compresses the image using sine/cosine approximations, taking less spaces on your drive to store the image.


Image files are stored in the file system of your computer. The location of it is specified using a "path", which is often unique. You can find the path of your current working directory with Python's `os` module. The `os` module provides functions to interact with the file system, e.g. creating or removing a directory (folder), listing its contents, changing and identifying the current working directory. 


```python
import os  # import module os để làm việc với hệ thống file
cwd = os.getcwd()  # lấy đường dẫn thư mục hiện tại
cwd  # hiển thị đường dẫn
```

## Load in Image in Python


OpenCV is a library used for computer vision. It has more functionality than the `PIL` library but is more difficult to use. We can import `OpenCV` as follows:


```python
import cv2  # import thư viện opencv
```

The <code>imread()</code> method loads an image from the specified file, the input is the <code>path</code> of the image to be read (just like PIL), the <code>flag</code> paramter specifies how the image should be read, and the default value is <code>cv2.IMREAD_COLOR</code>.


```python
image = cv2.imread(my_image)  # đọc ảnh từ file, trả về numpy array (định dạng BGR)
```

The result is a numpy array with intensity values as 8-bit unsigned integers. 


```python
type(image)  # kiểm tra kiểu dữ liệu (sẽ là numpy.ndarray)
```

We can get the shape of the array from the `shape` attribute.


```python
image.shape  # in shape của mảng (height, width, channels)
```

The shape is the same as the PIL array, but there are several differences; for example, PIL returns in (R, G, B) format whereas OpenCV returns in (B, G, R) format.


Each pixel could take on 256 possible values as intensity, ranging from 0 to 255, with 0 being the lowest intensity and 255 being the highest. The maximum and minimum intensity values of an image can be obtained, respectively, by calling:


```python
image.max()  # tìm giá trị pixel lớn nhất trong ảnh
```

and


```python
image.min()  # tìm giá trị pixel nhỏ nhất trong ảnh
```

### Grayscale Images


Grayscale images have pixel values representing the amount of light or intensity. Light shades of gray have a high-intensity darker shades have a lower intensity. White has the highest intensity, and black the lowest. We can convert an image to Gray Scale using a color conversion code and the function <code>cvtColor</code>.


The code for RGB to gray is <code>cv2.COLOR_BGR2GRAY</code>, we apply the function:


```python
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)  # chuyển ảnh sang grayscale
```

The image array has only two dimensions, i.e. only one color channel:


```python
image_gray.shape  # in shape của ảnh grayscale (chỉ có 2 chiều: height, width)
```

We can plot the image using `imshow` but we have to specify the color map is gray:


```python
plt.figure(figsize=(10, 10))  # tạo figure 10x10
plt.imshow(image_gray, cmap='gray')  # hiển thị ảnh grayscale với colormap gray
plt.show()  # hiển thị
```

We can save the image as a grayscale image, let's save it as a `jpg` as well, in the working directory.


```python
cv2.imwrite('lena_gray_cv.jpg', image_gray)  # lưu ảnh grayscale dưới định dạng JPG
```

You can also load in a grayscale image we have to set <code>flag</code> parameter to gray color conversation code: <code>cv2.COLOR_BGR2GRAY</code>:


```python
im_gray = cv2.imread('barbara.png', cv2.IMREAD_GRAYSCALE)  # đọc ảnh trực tiếp ở dạng grayscale
```

We can plot the image:


```python
plt.figure(figsize=(10,10))  # tạo figure 10x10
plt.imshow(im_gray,cmap='gray')  # hiển thị ảnh grayscale
plt.show()  # hiển thị
```

### Color Channels  


We can also work with the different color channels. Consider the following image:


```python
baboon=cv2.imread('baboon.png')  # đọc ảnh baboon
plt.figure(figsize=(10,10))  # tạo figure 10x10
plt.imshow(cv2.cvtColor(baboon, cv2.COLOR_BGR2RGB))  # chuyển BGR sang RGB và hiển thị
plt.show()  # hiển thị
```

We can obtain the different RGB colors and assign them to the variables <code>blue</code>, <code>green</code>, and <code>red</code>, in (B, G, R) format.


```python
blue, green, red = baboon[:, :, 0], baboon[:, :, 1], baboon[:, :, 2]  # tách 3 kênh màu BGR (blue=0, green=1, red=2)
```

We can concatenate each image channel the images using the function <code>vconcat</code>.


```python
im_bgr = cv2.vconcat([blue, green, red])  # nối 3 kênh màu theo chiều dọc
```

Plotting the color image next to the red channel in grayscale, we see that regions with red have higher intensity values.


```python
plt.figure(figsize=(10,10))  # tạo figure 10x10
plt.subplot(121)  # tạo subplot bên trái
plt.imshow(cv2.cvtColor(baboon, cv2.COLOR_BGR2RGB))  # hiển thị ảnh RGB
plt.title("RGB image")  # đặt tiêu đề
plt.subplot(122)  # tạo subplot bên phải
plt.imshow(im_bgr,cmap='gray')  # hiển thị các kênh màu dưới dạng grayscale
plt.title("Different color channels  blue (top), green (middle), red (bottom)  ")  # đặt tiêu đề
plt.show()  # hiển thị
```

### Indexing  


We can use numpy slicing. For example, we can return the first 256 rows corresponding to the top half of the image:


```python
rows = 256  # số hàng muốn lấy (nửa trên của ảnh)
```

```python
plt.figure(figsize=(10,10))  # tạo figure 10x10
plt.imshow(new_image[0:rows,:,:])  # hiển thị nửa trên của ảnh (từ hàng 0 đến rows)
plt.show()  # hiển thị
```

We can also return the first 256 columns corresponding to the first half of the image:


```python
columns = 256  # số cột muốn lấy (nửa trái của ảnh)
```

```python
plt.figure(figsize=(10,10))  # tạo figure 10x10
plt.imshow(new_image[:,0:columns,:])  # hiển thị nửa trái của ảnh (từ cột 0 đến columns)
plt.show()  # hiển thị
```

If you want to reassign an array to another variable, you should use the `copy` method (we will cover this in the next section).


```python
A = new_image.copy()  # tạo bản sao độc lập của mảng
plt.imshow(A)  # hiển thị ảnh
plt.show()  # hiển thị
```

If we do not apply the method `copy()`, the variable will point to the same location in memory. Consider the variable `B` below, if we set all values of array `A` to zero, since `A` and `B` points to the same object in the memory, `B` will also have all-zero elements:


```python
B = A  # gán B trỏ đến cùng vị trí bộ nhớ với A
A[:,:,:] = 0  # đặt tất cả giá trị của A về 0
plt.imshow(B)  # hiển thị B (cũng sẽ là 0 vì cùng bộ nhớ)
plt.show()  # hiển thị
```

We can also manipulate elements using indexing. In the following piece of code, we create a new array `baboon_red` and set all but the red color channels to zero. Therefore, when we display the image, it appears red:


```python
baboon_red = baboon.copy()  # tạo bản sao của baboon
baboon_red[:, :, 0] = 0  # đặt kênh blue về 0
baboon_red[:, :, 1] = 0  # đặt kênh green về 0
plt.figure(figsize=(10, 10))  # tạo figure 10x10
plt.imshow(cv2.cvtColor(baboon_red, cv2.COLOR_BGR2RGB))  # chuyển BGR sang RGB và hiển thị chỉ kênh màu đỏ
plt.show()  # hiển thị
```

```python
baboon_blue = baboon.copy()  # tạo bản sao của baboon
baboon_blue[:, :, 1] = 0  # đặt kênh green về 0
baboon_blue[:, :, 2] = 0  # đặt kênh red về 0
plt.figure(figsize=(10, 10))  # tạo figure 10x10
plt.imshow(cv2.cvtColor(baboon_blue, cv2.COLOR_BGR2RGB))  # chuyển BGR sang RGB và hiển thị chỉ kênh màu xanh dương
plt.show()  # hiển thị
```

## So sánh PIL (Pillow) và OpenCV

### 1. Định dạng màu
- **PIL**: Sử dụng định dạng **RGB** (Red-Green-Blue)
- **OpenCV**: Sử dụng định dạng **BGR** (Blue-Green-Red)
- **Lưu ý**: Khi hiển thị ảnh OpenCV với matplotlib, cần chuyển đổi từ BGR sang RGB bằng `cv2.cvtColor(image, cv2.COLOR_BGR2RGB)`

### 2. Kiểu dữ liệu trả về
- **PIL**: Trả về đối tượng `PIL.Image` - cần chuyển đổi sang NumPy array bằng `np.array()` hoặc `np.asarray()`
- **OpenCV**: Trả về trực tiếp **NumPy array** - không cần chuyển đổi

### 3. Indexing kênh màu
- **PIL**: `array[:, :, 0]` = Red, `array[:, :, 1]` = Green, `array[:, :, 2]` = Blue
- **OpenCV**: `array[:, :, 0]` = Blue, `array[:, :, 1]` = Green, `array[:, :, 2]` = Red

### 4. Độ phức tạp và dễ sử dụng
- **PIL**: 
  - Đơn giản, dễ học và sử dụng
  - API thân thiện với người mới bắt đầu
  - Phù hợp cho các tác vụ xử lý ảnh cơ bản
  
- **OpenCV**: 
  - Nhiều tính năng hơn, phức tạp hơn
  - API đầy đủ cho computer vision
  - Phù hợp cho các dự án nâng cao

### 5. Tốc độ xử lý
- **PIL**: Chậm hơn, phù hợp cho xử lý ảnh đơn lẻ
- **OpenCV**: Nhanh hơn, được tối ưu cho xử lý ảnh và video real-time

### 6. Tính năng
- **PIL**: 
  - Xử lý ảnh cơ bản (mở, lưu, resize, crop, rotate)
  - Chuyển đổi định dạng
  - Quantization
  - Lọc ảnh cơ bản
  
- **OpenCV**: 
  - Tất cả tính năng của PIL
  - Computer vision nâng cao (object detection, face recognition)
  - Xử lý video
  - Machine learning tích hợp
  - Camera và video capture

### 7. Cách lưu ảnh
- **PIL**: `image.save('output.jpg')` - sử dụng method của Image object
- **OpenCV**: `cv2.imwrite('output.jpg', image)` - sử dụng function với array

### 8. Chuyển đổi grayscale
- **PIL**: `ImageOps.grayscale(image)` - trả về PIL Image với mode 'L'
- **OpenCV**: `cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)` - trả về NumPy array 2 chiều

### 9. Tách kênh màu
- **PIL**: `red, green, blue = image.split()` - trả về 3 PIL Image objects
- **OpenCV**: `blue, green, red = image[:, :, 0], image[:, :, 1], image[:, :, 2]` - trả về 3 NumPy arrays

### Kết luận
- **Chọn PIL**: khi cần xử lý ảnh đơn giản, dễ học, không cần tốc độ cao
- **Chọn OpenCV**: khi cần tính năng computer vision nâng cao, xử lý video, tốc độ cao, hoặc làm việc với camera
